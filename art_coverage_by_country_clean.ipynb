```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Load dataset
file_path = "art_coverage_by_country_clean.csv"
df = pd.read_csv(file_path)

# Drop rows with missing values
df.dropna(inplace=True)

# Encoding categorical variables if any
label_encoders = {}
for col in df.select_dtypes(include=['object']).columns:
    label_encoders[col] = LabelEncoder()
    df[col] = label_encoders[col].fit_transform(df[col])

# Splitting features and target variable (assuming last column is target)
X = df.iloc[:, :-1]
y = df.iloc[:, -1]

# Standardizing the features
scaler = StandardScaler()
X = scaler.fit_transform(X)

# Function to evaluate SVM model
def evaluate_svm(train_size):
    X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=train_size, random_state=42)
    model = SVC(kernel='linear')
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, average='weighted', zero_division=1)
    recall = recall_score(y_test, y_pred, average='weighted', zero_division=1)
    f1 = f1_score(y_test, y_pred, average='weighted', zero_division=1)

    print(f"Train Size: {train_size} | Accuracy: {accuracy:.4f} | Precision: {precision:.4f} | Recall: {recall:.4f} | F1 Score: {f1:.4f}")

# Evaluating for different train-test splits
for ratio in [0.8, 0.6, 0.5]:
    evaluate_svm(ratio)
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import StackingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Load dataset
file_path = "art_coverage_by_country_clean.csv"
df = pd.read_csv(file_path)

# Drop rows with missing values
df.dropna(inplace=True)

# Encoding categorical variables if any
label_encoders = {}
for col in df.select_dtypes(include=['object']).columns:
    label_encoders[col] = LabelEncoder()
    df[col] = label_encoders[col].fit_transform(df[col])

# Splitting features and target variable (assuming last column is target)
X = df.iloc[:, :-1]
y = df.iloc[:, -1]

# Standardizing the features
scaler = StandardScaler()
X = scaler.fit_transform(X)

# Define base models for stacking
base_models = [
    ('log_reg', LogisticRegression(max_iter=1000)),
    ('decision_tree', DecisionTreeClassifier()),
    ('svm', SVC(probability=True))
]

# Define meta model
meta_model = RandomForestClassifier()

# Create stacking classifier
stacking_clf = StackingClassifier(estimators=base_models, final_estimator=meta_model, passthrough=True)

# Function to evaluate Stacking model
def evaluate_stacking(train_size):
    X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=train_size, random_state=42)
    stacking_clf.fit(X_train, y_train)
    y_pred = stacking_clf.predict(X_test)

    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, average='weighted', zero_division=1)
    recall = recall_score(y_test, y_pred, average='weighted', zero_division=1)
    f1 = f1_score(y_test, y_pred, average='weighted', zero_division=1)

    print(f"Train Size: {train_size} | Accuracy: {accuracy:.4f} | Precision: {precision:.4f} | Recall: {recall:.4f} | F1 Score: {f1:.4f}")

# Evaluating for different train-test splits
for ratio in [0.8, 0.6, 0.5]:
    evaluate_stacking(ratio)
```

```python
import pennylane as qml
import torch
import torch.nn as nn
import torch.optim as optim
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.decomposition import PCA
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Quantum device setup
n_qubits = 4
dev = qml.device("default.qubit", wires=n_qubits)

# Define Quantum Neural Network
@qml.qnode(dev, interface="torch")
def quantum_circuit(inputs, weights):
    qml.AngleEmbedding(inputs, wires=range(n_qubits), rotation="Y")
    qml.StronglyEntanglingLayers(weights, wires=range(n_qubits))
    return [qml.expval(qml.PauliZ(i)) for i in range(n_qubits)]

# PyTorch Model
class QNN(nn.Module):
    def __init__(self, output_size):
        super().__init__()
        weight_shapes = {"weights": (3, n_qubits, 3)}
        self.q_layer = qml.qnn.TorchLayer(quantum_circuit, weight_shapes)
        self.fc = nn.Linear(n_qubits, output_size)

    def forward(self, x):
        x = self.q_layer(x)
        return self.fc(x)

# Load and preprocess data
file_path = "art_coverage_by_country_clean.csv"
df = pd.read_csv(file_path)
df.dropna(inplace=True)

# Encode categorical variables
label_encoders = {}
for col in df.select_dtypes(include=['object']).columns:
    label_encoders[col] = LabelEncoder()
    df[col] = label_encoders[col].fit_transform(df[col])

# Feature selection and dimensionality reduction using PCA
scaler = StandardScaler()
X = scaler.fit_transform(df.iloc[:, :-1])  # Exclude target column

pca = PCA(n_components=4)  # Reduce to 4 features for QNN
X = pca.fit_transform(X)

# Convert target into categorical bins using `pd.cut`
df['target_class'] = pd.cut(df.iloc[:, -1], bins=5, labels=False, duplicates='drop')

y = df['target_class']

# Train and evaluate QNN
def train_and_evaluate_qnn(X, y, train_size):
    X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=train_size, random_state=42)

    # Convert to PyTorch tensors
    X_train, X_test = torch.tensor(X_train, dtype=torch.float32), torch.tensor(X_test, dtype=torch.float32)
    y_train, y_test = torch.tensor(y_train.values, dtype=torch.long), torch.tensor(y_test.values, dtype=torch.long)

    # Initialize model, loss, and optimizer
    model = QNN(output_size=len(np.unique(y)))
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.Adam(model.parameters(), lr=0.01)

    # Training loop
    epochs = 20
    for epoch in range(epochs):
        optimizer.zero_grad()
        outputs = model(X_train)
        loss = criterion(outputs, y_train)
        loss.backward()
        optimizer.step()

    # Predictions
    with torch.no_grad():
        y_pred = model(X_test).argmax(dim=1)

    # Calculate Metrics
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, average='weighted', zero_division=1)
    recall = recall_score(y_test, y_pred, average='weighted', zero_division=1)
    f1 = f1_score(y_test, y_pred, average='weighted', zero_division=1)

    print(f"[Art Coverage] Train Size: {train_size} | Accuracy: {accuracy:.4f} | Precision: {precision:.4f} | Recall: {recall:.4f} | F1 Score: {f1:.4f}")

# Train and evaluate for different splits
for ratio in [0.8, 0.6, 0.5]:
    train_and_evaluate_qnn(X, y, ratio)
```

```python
import torch
import torch.nn.functional as F
from torch_geometric.data import Data
from torch_geometric.nn import GCNConv
from torch_geometric.utils import dense_to_sparse
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import pandas as pd
import numpy as np
import re

# Load dataset
file_path = "art_coverage_by_country_clean.csv"
df = pd.read_csv(file_path)

# Drop non-numeric column ('Country')
df.drop(columns=['Country'], inplace=True)

# Function to clean numeric values
def extract_numeric(value):
    """ Extracts the first numeric value from strings like '7200[4100–11000]' """
    if isinstance(value, str):
        match = re.search(r"\d+", value)
        return float(match.group()) if match else np.nan
    return value

# Apply cleaning
df = df.applymap(extract_numeric)

# Handle missing values
df.fillna(df.median(numeric_only=True), inplace=True)

# Encode categorical target ('WHO Region')
label_encoder = LabelEncoder()
df['WHO Region'] = label_encoder.fit_transform(df['WHO Region'])

# Define features (X) and target (y)
X = df.drop(columns=['WHO Region'])
y = df['WHO Region']

# Standardize features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Convert to PyTorch tensors
X_tensor = torch.tensor(X_scaled, dtype=torch.float)
y_tensor = torch.tensor(y.values, dtype=torch.long)

# Ensure target labels are within bounds
num_classes = len(torch.unique(y_tensor))
y_tensor = torch.clamp(y_tensor, 0, num_classes - 1)

# Define a GNN Model
class GNNModel(torch.nn.Module):
    def __init__(self, input_dim, hidden_dim, output_dim):
        super(GNNModel, self).__init__()
        self.conv1 = GCNConv(input_dim, hidden_dim)
        self.conv2 = GCNConv(hidden_dim, output_dim)

    def forward(self, x, edge_index):
        x = self.conv1(x, edge_index)
        x = F.relu(x)
        x = self.conv2(x, edge_index)
        return F.log_softmax(x, dim=1)

# Function to Train & Evaluate GNN
def train_and_evaluate_gnn(X_tensor, y_tensor, train_size):
    X_train, X_test, y_train, y_test = train_test_split(X_tensor, y_tensor, train_size=train_size, random_state=42, stratify=y_tensor)

    # Create adjacency matrix for training data only
    num_train_nodes = X_train.shape[0]
    adj_matrix = torch.eye(num_train_nodes)  # Self-loop adjacency matrix (identity matrix)
    edge_index, _ = dense_to_sparse(adj_matrix)

    # Create Data Object
    data = Data(x=X_train, edge_index=edge_index, y=y_train)

    model = GNNModel(input_dim=X_tensor.shape[1], hidden_dim=16, output_dim=num_classes)
    optimizer = torch.optim.Adam(model.parameters(), lr=0.01)

    # Training
    model.train()
    for epoch in range(100):
        optimizer.zero_grad()
        out = model(data.x, data.edge_index)
        loss = F.nll_loss(out, data.y)
        loss.backward()
        optimizer.step()

    # Evaluation
    model.eval()
    with torch.no_grad():
        test_out = model(X_test, edge_index)  # Edge index now properly constrained
        y_pred = test_out.argmax(dim=1).cpu().numpy()
        y_test_np = y_test.cpu().numpy()

    # Calculate Metrics
    accuracy = accuracy_score(y_test_np, y_pred)
    precision = precision_score(y_test_np, y_pred, average='weighted', zero_division=1)
    recall = recall_score(y_test_np, y_pred, average='weighted', zero_division=1)
    f1 = f1_score(y_test_np, y_pred, average='weighted', zero_division=1)

    return accuracy, precision, recall, f1

# Run for different train-test splits
for ratio in [0.8, 0.6, 0.5]:
    acc, prec, rec, f1 = train_and_evaluate_gnn(X_tensor, y_tensor, ratio)
    print(f"Train Ratio {ratio}: Accuracy={acc:.4f}, Precision={prec:.4f}, Recall={rec:.4f}, F1-Score={f1:.4f}")
```

```python
import torch
import torch.nn.functional as F
from torch_geometric.data import Data
from torch_geometric.nn import GCNConv
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.neighbors import NearestNeighbors
import pandas as pd
import numpy as np
import re

# Load dataset
file_path = "art_coverage_by_country_clean.csv"
df = pd.read_csv(file_path)

# Drop non-numeric column ('Country')
df.drop(columns=['Country'], inplace=True)

# Function to clean numeric values
def extract_numeric(value):
    if isinstance(value, str):
        match = re.search(r"\d+", value)
        return float(match.group()) if match else np.nan
    return value

df = df.applymap(extract_numeric)
df.fillna(df.median(numeric_only=True), inplace=True)

# Encode categorical target ('WHO Region')
label_encoder = LabelEncoder()
df['WHO Region'] = label_encoder.fit_transform(df['WHO Region'])

# Features (X) and target (y)
X = df.drop(columns=['WHO Region'])
y = df['WHO Region']

# Standardize features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Convert to PyTorch tensors
X_tensor = torch.tensor(X_scaled, dtype=torch.float)
y_tensor = torch.tensor(y.values, dtype=torch.long)

# Ensure target labels are valid
num_classes = len(torch.unique(y_tensor))
y_tensor = torch.clamp(y_tensor, 0, num_classes - 1)

# Function to construct edge index using nearest neighbors
def create_edge_index(X_tensor, k=5):
    nn = NearestNeighbors(n_neighbors=k, metric='euclidean')
    nn.fit(X_tensor.numpy())
    edges = nn.kneighbors(X_tensor.numpy(), return_distance=False)
    edge_index = torch.tensor([[i, j] for i, row in enumerate(edges) for j in row if i != j], dtype=torch.long).t()
    return edge_index

# Define GNN Model
class GNNModel(torch.nn.Module):
    def __init__(self, input_dim, hidden_dim, output_dim):
        super(GNNModel, self).__init__()
        self.conv1 = GCNConv(input_dim, hidden_dim)
        self.conv2 = GCNConv(hidden_dim, output_dim)

    def forward(self, x, edge_index):
        x = self.conv1(x, edge_index)
        x = F.relu(x)
        x = self.conv2(x, edge_index)
        return F.log_softmax(x, dim=1)

# Train & Evaluate GNN
def train_and_evaluate_gnn(X_tensor, y_tensor, train_size):
    X_train, X_test, y_train, y_test = train_test_split(X_tensor, y_tensor, train_size=train_size, random_state=42, stratify=y_tensor)

    # Create edge index using Nearest Neighbors
    edge_index = create_edge_index(X_train)

    data = Data(x=X_train, edge_index=edge_index, y=y_train)
    model = GNNModel(input_dim=X_tensor.shape[1], hidden_dim=16, output_dim=num_classes)
    optimizer = torch.optim.Adam(model.parameters(), lr=0.01)

    # Train model
    model.train()
    for epoch in range(100):
        optimizer.zero_grad()
        out = model(data.x, data.edge_index)
        loss = F.nll_loss(out, data.y)
        loss.backward()
        optimizer.step()

    # Evaluate
    model.eval()
    with torch.no_grad():
        test_edge_index = create_edge_index(X_test)
        test_out = model(X_test, test_edge_index)
        y_pred = test_out.argmax(dim=1).cpu().numpy()
        y_test_np = y_test.cpu().numpy()

    # Metrics
    accuracy = accuracy_score(y_test_np, y_pred)
    precision = precision_score(y_test_np, y_pred, average='weighted', zero_division=1)
    recall = recall_score(y_test_np, y_pred, average='weighted', zero_division=1)
    f1 = f1_score(y_test_np, y_pred, average='weighted', zero_division=1)

    return accuracy, precision, recall, f1

# Run for different train-test splits
for ratio in [0.8, 0.6, 0.5]:
    acc, prec, rec, f1 = train_and_evaluate_gnn(X_tensor, y_tensor, ratio)
    print(f"Train Ratio {ratio}: Accuracy={acc:.4f}, Precision={prec:.4f}, Recall={rec:.4f}, F1-Score={f1:.4f}")
```
