```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Load dataset
file_path = "annual-number-of-deaths-by-cause.csv"
df = pd.read_csv(file_path)

# Drop rows with missing values
df.dropna(inplace=True)

# Encoding categorical variables if any
label_encoders = {}
for col in df.select_dtypes(include=['object']).columns:
    label_encoders[col] = LabelEncoder()
    df[col] = label_encoders[col].fit_transform(df[col])

# Splitting features and target variable (assuming last column is target)
X = df.iloc[:, :-1]
y = df.iloc[:, -1]

# Standardizing the features
scaler = StandardScaler()
X = scaler.fit_transform(X)

# Function to evaluate SVM model
def evaluate_svm(train_size):
    X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=train_size, random_state=42)
    model = SVC(kernel='linear')
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, average='weighted', zero_division=1)
    recall = recall_score(y_test, y_pred, average='weighted', zero_division=1)
    f1 = f1_score(y_test, y_pred, average='weighted', zero_division=1)

    print(f"Train Size: {train_size} | Accuracy: {accuracy:.4f} | Precision: {precision:.4f} | Recall: {recall:.4f} | F1 Score: {f1:.4f}")

# Evaluating for different train-test splits
for ratio in [0.8, 0.6, 0.5]:
    evaluate_svm(ratio)
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import StackingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Load dataset
file_path = "annual-number-of-deaths-by-cause.csv"
df = pd.read_csv(file_path)

# Drop rows with missing values
df.dropna(inplace=True)

# Encoding categorical variables if any
label_encoders = {}
for col in df.select_dtypes(include=['object']).columns:
    label_encoders[col] = LabelEncoder()
    df[col] = label_encoders[col].fit_transform(df[col])

# Splitting features and target variable (assuming last column is target)
X = df.iloc[:, :-1]
y = df.iloc[:, -1]

# Standardizing the features
scaler = StandardScaler()
X = scaler.fit_transform(X)

# Define base models for stacking
base_models = [
    ('log_reg', LogisticRegression(max_iter=1000)),
    ('decision_tree', DecisionTreeClassifier()),
    ('svm', SVC(probability=True))
]

# Define meta model
meta_model = RandomForestClassifier()

# Create stacking classifier with reduced cross-validation folds
stacking_clf = StackingClassifier(estimators=base_models, final_estimator=meta_model, passthrough=True, cv=3)  # Changed cv=5 to cv=3

# Function to evaluate Stacking model
def evaluate_stacking(train_size):
    X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=train_size, random_state=42)
    stacking_clf.fit(X_train, y_train)
    y_pred = stacking_clf.predict(X_test)

    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, average='weighted', zero_division=1)
    recall = recall_score(y_test, y_pred, average='weighted', zero_division=1)
    f1 = f1_score(y_test, y_pred, average='weighted', zero_division=1)

    print(f"Train Size: {train_size} | Accuracy: {accuracy:.4f} | Precision: {precision:.4f} | Recall: {recall:.4f} | F1 Score: {f1:.4f}")

# Evaluating for different train-test splits
for ratio in [0.8, 0.6, 0.5]:
    evaluate_stacking(ratio)
```

```python
import pennylane as qml
import torch
import torch.nn as nn
import torch.optim as optim
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.decomposition import PCA
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Quantum device setup
n_qubits = 4
dev = qml.device("default.qubit", wires=n_qubits)

# Define Quantum Neural Network
@qml.qnode(dev, interface="torch")
def quantum_circuit(inputs, weights):
    qml.AngleEmbedding(inputs, wires=range(n_qubits), rotation="Y")
    qml.StronglyEntanglingLayers(weights, wires=range(n_qubits))
    return [qml.expval(qml.PauliZ(i)) for i in range(n_qubits)]

# PyTorch Model
class QNN(nn.Module):
    def __init__(self, output_size):
        super().__init__()
        weight_shapes = {"weights": (3, n_qubits, 3)}
        self.q_layer = qml.qnn.TorchLayer(quantum_circuit, weight_shapes)
        self.fc = nn.Linear(n_qubits, output_size)

    def forward(self, x):
        x = self.q_layer(x)
        return self.fc(x)

# Load and preprocess data
file_path = "annual-number-of-deaths-by-cause.csv"
df = pd.read_csv(file_path)
df.dropna(inplace=True)

# Encode categorical variables
label_encoders = {}
for col in df.select_dtypes(include=['object']).columns:
    label_encoders[col] = LabelEncoder()
    df[col] = label_encoders[col].fit_transform(df[col])

# Feature selection and dimensionality reduction using PCA
scaler = StandardScaler()
X = scaler.fit_transform(df.iloc[:, :-1])  # Exclude target column

pca = PCA(n_components=4)  # Reduce to 4 features for QNN
X = pca.fit_transform(X)

# Convert target into categorical bins ensuring non-empty bins
df['target_class'], bins = pd.cut(df.iloc[:, -1], bins=5, labels=False, retbins=True, duplicates='drop')

# Fix target labels to be continuous starting from 0
df['target_class'] = pd.factorize(df['target_class'])[0]

y = df['target_class'].astype(int)  # Ensure integer labels

# Train and evaluate QNN
def train_and_evaluate_qnn(X, y, train_size):
    X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=train_size, random_state=42)

    unique_classes = np.unique(y_train)  # Ensure correct number of output neurons
    output_size = len(unique_classes)

    # Convert to PyTorch tensors
    X_train, X_test = torch.tensor(X_train, dtype=torch.float32), torch.tensor(X_test, dtype=torch.float32)
    y_train, y_test = torch.tensor(y_train.values, dtype=torch.long), torch.tensor(y_test.values, dtype=torch.long)

    # Initialize model, loss, and optimizer
    model = QNN(output_size=output_size)
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.Adam(model.parameters(), lr=0.01)

    # Training loop
    epochs = 20
    for epoch in range(epochs):
        optimizer.zero_grad()
        outputs = model(X_train)
        loss = criterion(outputs, y_train)
        loss.backward()
        optimizer.step()

    # Predictions
    with torch.no_grad():
        y_pred = model(X_test).argmax(dim=1)

    # Calculate Metrics
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, average='weighted', zero_division=1)
    recall = recall_score(y_test, y_pred, average='weighted', zero_division=1)
    f1 = f1_score(y_test, y_pred, average='weighted', zero_division=1)

    print(f"[Annual Deaths] Train Size: {train_size} | Accuracy: {accuracy:.4f} | Precision: {precision:.4f} | Recall: {recall:.4f} | F1 Score: {f1:.4f}")

# Train and evaluate for different splits
for ratio in [0.8, 0.6, 0.5]:
    train_and_evaluate_qnn(X, y, ratio)
```

```python
import torch
import torch.nn.functional as F
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.decomposition import PCA
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from torch_geometric.data import Data
from torch_geometric.nn import GCNConv

# Load Dataset
file_path = "annual-number-of-deaths-by-cause.csv"
df = pd.read_csv(file_path)
df.dropna(inplace=True)

# Encode categorical variables
label_encoders = {}
for col in df.select_dtypes(include=['object']).columns:
    label_encoders[col] = LabelEncoder()
    df[col] = label_encoders[col].fit_transform(df[col])

# Feature Scaling & Dimensionality Reduction
scaler = StandardScaler()
X = scaler.fit_transform(df.iloc[:, :-1])  # Exclude target column
pca = PCA(n_components=4)  # Reduce to 4 dimensions
X = pca.fit_transform(X)

# Convert target column into discrete categories
df['target_class'], bins = pd.cut(df.iloc[:, -1], bins=5, labels=False, retbins=True, duplicates='drop')
df['target_class'] = pd.factorize(df['target_class'])[0]  # Ensure continuous labels
y = df['target_class'].astype(int)

# Create Graph Structure (Edges based on similarity)
num_samples = len(X)
edge_index = torch.tensor([(i, j) for i in range(num_samples) for j in range(i+1, num_samples) if np.linalg.norm(X[i] - X[j]) < 1.0], dtype=torch.long).T

# Convert to PyTorch tensors
X_tensor = torch.tensor(X, dtype=torch.float32)
y_tensor = torch.tensor(y.values, dtype=torch.long)

# Create Graph Data Object
graph_data = Data(x=X_tensor, edge_index=edge_index, y=y_tensor)

# Define GNN Model
class GNN(torch.nn.Module):
    def __init__(self, input_dim, output_dim):
        super(GNN, self).__init__()
        self.conv1 = GCNConv(input_dim, 16)
        self.conv2 = GCNConv(16, output_dim)

    def forward(self, data):
        x, edge_index = data.x, data.edge_index
        x = self.conv1(x, edge_index)
        x = F.relu(x)
        x = self.conv2(x, edge_index)
        return F.log_softmax(x, dim=1)

# Train and Evaluate
def train_and_evaluate_gnn(graph_data, train_size):
    num_samples = graph_data.x.shape[0]
    train_mask = torch.zeros(num_samples, dtype=torch.bool)
    train_idx, test_idx = train_test_split(range(num_samples), train_size=train_size, random_state=42)
    train_mask[train_idx] = True

    model = GNN(input_dim=4, output_dim=len(torch.unique(y_tensor)))
    optimizer = torch.optim.Adam(model.parameters(), lr=0.01)
    criterion = torch.nn.CrossEntropyLoss()

    # Training Loop
    for epoch in range(20):
        optimizer.zero_grad()
        out = model(graph_data)
        loss = criterion(out[train_mask], y_tensor[train_mask])
        loss.backward()
        optimizer.step()

    # Predictions
    y_pred = out.argmax(dim=1)
    y_test, y_pred = y_tensor[~train_mask], y_pred[~train_mask]

    # Compute Metrics
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, average='weighted', zero_division=1)
    recall = recall_score(y_test, y_pred, average='weighted', zero_division=1)
    f1 = f1_score(y_test, y_pred, average='weighted', zero_division=1)

    print(f"[Annual Deaths] Train: {train_size} | Acc: {accuracy:.4f} | Prec: {precision:.4f} | Recall: {recall:.4f} | F1: {f1:.4f}")

# Run for different splits
for ratio in [0.8, 0.6, 0.5]:
    train_and_evaluate_gnn(graph_data, ratio)
```
